```python
# Build the circuit for Double-Slit Interference (Quantum Interference)
from qiskit import QuantumCircuit, transpile
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.h(0) # Superposition: particle goes through both slits
qc.rz(3.1415 / 2, 0) # Phase shift: simulating path difference
qc.h(0) # Interference
qc.measure(0, 0) # Measurement
qc.draw(output="mpl")
```

```python
# Simulation on AerSimulator
simulator = AerSimulator() # Instantiate the AerSimulator
transpiled_qc = transpile(qc, simulator) # Transpile the circuit for the simulator
job = simulator.run(transpiled_qc, shots=1024)  # Run with 1024 shots

result=job.result()
counts = result.get_counts()
plot_histogram(counts)
```

```python
# Build the circuit for Schrödinger’s Cat experiment
qc = QuantumCircuit(2, 2)
qc.h(0) # Cat in superposition: alive and dead
qc.cx(0, 1) # Entangle with poison mechanism
qc.measure([0, 1], [0, 1]) # Measure both qubits
qc.draw("mpl")
```

```python
# Simulation on AerSimulator
simulator = AerSimulator() # Instantiate the AerSimulator
transpiled_qc = transpile(qc, simulator) # Transpile the circuit for the simulator
job = simulator.run(transpiled_qc, shots=1024)  # Run with 1024 shots
result=job.result()

counts = result.get_counts()
plot_histogram(counts)
```

```python
# Build the circuit for Bell Inequality (Quantum Entanglement)
qc = QuantumCircuit(2, 2)
qc.h(0) # Superposition
qc.cx(0, 1) # Create Bell state
qc.measure([0, 1], [0, 1]) # Measure both qubits
qc.draw("mpl")
# from the circuit, we can find there is no difference to Schrödinger’s Cat experiment, which suggests Schrödinger’s Cat is also an entanglement phenomenon.
```

```python
# Simulation on AerSimulator
simulator = AerSimulator() # Instantiate the AerSimulator
transpiled_qc = transpile(qc, simulator) # Transpile the circuit for the simulator
job = simulator.run(transpiled_qc, shots=1024)  # Run with 1024 shots
result=job.result()

counts = result.get_counts()
plot_histogram(counts)
```

```python
# Build the circuit for Aharonov-Bohm (AB) Effect Simulation
# To simulate the phase shift caused by a vector potential, even when the particle does not pass through a magnetic field—mimicking the AB effect.

qc = QuantumCircuit(1, 1)
qc.h(0) # Superposition: two paths
qc.rz(3.1415 / 4, 0) # Phase shift: simulating vector potential
qc.h(0) # Interference
qc.measure(0, 0)
qc.draw("mpl")
```

```python
# Simulation on AerSimulator
simulator = AerSimulator() # Instantiate the AerSimulator
transpiled_qc = transpile(qc, simulator) # Transpile the circuit for the simulator
job = simulator.run(transpiled_qc, shots=1024)  # Run with 1024 shots
result=job.result()

counts = result.get_counts()
plot_histogram(counts)
# Aharonov-Bohm (AB) effect simulation behaves like a quantum interference setup, but with a phase shift that biases the outcome toward |1⟩
# why? because the probability of measuring |1⟩ is (1 - cosθ)/2
```
