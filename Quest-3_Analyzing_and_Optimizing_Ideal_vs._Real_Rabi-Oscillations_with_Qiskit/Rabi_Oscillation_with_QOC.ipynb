```python
#Quantum Optimal Control
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_ibm_runtime.fake_provider import FakePerth
from qiskit_aer import AerSimulator
from scipy.optimize import minimize

# 1. Select the noisy backend
backend = AerSimulator.from_backend(FakePerth())

# 2. Define the cost function
def cost_function(angle):
    """
    This function runs a simulation with a specified Rx angle and returns the cost.
    Our goal is to minimize this cost.
    """
    # Create the quantum circuit
    qc = QuantumCircuit(1, 1)
    qc.rx(angle[0], 0)
    qc.measure(0, 0)

    # Transpile the circuit and run on the simulator
    job = backend.run(transpile(qc, backend), shots=1024)
    result = job.result()
    counts = result.get_counts()

    # Calculate the probability of measuring |1>
    prob_1 = counts.get('1', 0) / 1024
    
    # Cost function: we want this value to be as small as possible
    cost = 1 - prob_1
    print(f"Attempted Angle: {angle[0]:.4f} rad, Measured Prob: {prob_1:.4f}, Cost: {cost:.4f}")

    return cost

# 3. Execute the optimization
# We start with the ideal pi angle as our initial guess
initial_guess = [np.pi]

# Use the minimize function to find the best angle
result = minimize(
    cost_function,
    initial_guess,
    method='Powell', # An optimization method suitable for this problem
    options={'disp': True}
)

# 4. Display the results
optimized_angle = result.x[0]
optimized_prob = 1 - result.fun

print("\n--- Optimization Results ---")
print(f"Ideal pi Angle: {np.pi:.4f} rad")
print(f"Optimized Angle Found: {optimized_angle:.4f} rad")
print(f"Probability of measuring |1> at this optimized angle: {optimized_prob:.4f}")
```
